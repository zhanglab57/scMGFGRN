```python
import pandas as pd
import numpy as np
import os
df=pd.read_csv("AllPair.csv")
arr = df.values
print(arr)

seed = 0
os.environ['PYTHONHASHSEED'] = str(seed)

np.random.seed(seed)
np.random.shuffle(arr)

#df=pd.DataFrame(df)

print(arr)
test_index = np.load("test_index.npy")
train_index = np.load("train_index.npy")


test_set = arr[test_index]
train_set = arr[train_index]

# 计算验证集大小
n_samples = train_set.shape[0]
val_size = int(n_samples * 0.1)

# 选择前10%作为验证集
val_set = train_set[:val_size]
new_train_set = train_set[val_size:]


print(f"原始训练集大小: {train_set.shape}")
print(f"新训练集大小: {new_train_set.shape}")
print(f"验证集大小: {val_set.shape}")

test_set = pd.DataFrame(test_set)
train_set = pd.DataFrame(new_train_set)
val_set = pd.DataFrame(val_set)

test_set.to_csv("./tvt/test_set.csv",index=False)
test_set.to_csv("./tvt/train_set.csv",index=False)
test_set.to_csv("./tvt/val_set.csv",index=False)
```
